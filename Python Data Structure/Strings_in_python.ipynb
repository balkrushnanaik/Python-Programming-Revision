# Python Strings

A **string** is a sequence of characters enclosed inside **single quotes (`' '`), double quotes (`" "`), or triple quotes (`''' '''` / `""" """`)**.

```python
name = "Balkrushna"
city = 'Pune'
```

Both are valid strings.

## 1. Creating a String

```python
name = "Balkrushna"
message = 'Welcome to Python'
```

You can check the data type:

```python
print(type(name))
```

Output:

```text
<class 'str'>
```

---

## 2. Multiline String

Triple quotes are used for multiline strings.

```python
message = """Hello,
Welcome to Python.
Let's learn strings."""

print(message)
```

---

## 3. String Indexing

Each character has an index.

```python
text = "Python"
```

| Character      |  P |  y |  t |  h |  o |  n |
| -------------- | -: | -: | -: | -: | -: | -: |
| Positive Index |  0 |  1 |  2 |  3 |  4 |  5 |
| Negative Index | -6 | -5 | -4 | -3 | -2 | -1 |

Example:

```python
text = "Python"

print(text[0])
print(text[2])
print(text[-1])
```

Output:

```text
P
t
n
```

---

## 4. String Slicing

Slicing extracts a portion of a string.

### Syntax

```python
string[start:end]
```

The `end` index is excluded.

```python
text = "Python"

print(text[0:3])
print(text[2:6])
```

Output:

```text
Pyt
thon
```

### Other slicing examples

```python
text = "Python"

print(text[:3])      # Pyt
print(text[3:])      # hon
print(text[:])       # Python
print(text[::2])     # Pto
print(text[::-1])    # nohtyP
```

# 5. String Concatenation

Joining strings using `+` is called concatenation.

```python
first_name = "Balkrushna"
last_name = "Naik"

full_name = first_name + " " + last_name

print(full_name)
```

Output:

```text
Balkrushna Naik
```

---

# 6. String Repetition

You can repeat a string using `*`.

```python
text = "Python "

print(text * 3)
```

Output:

```text
Python Python Python
```

# 7. `len()` Function

`len()` returns the number of characters.

```python
text = "Python"

print(len(text))
```

Output:

```text
6
```

Spaces are also counted.

```python
text = "Hello World"

print(len(text))
```

Output:

```text
11
```

# 10. `capitalize()`

Converts the first character to uppercase.

```python

```

```text

```

---

# 11. `title()`

Capitalizes the first character of every word.

```python

print(text.title())
```

Output:

```text

```

---

# 12. `strip()`

Removes spaces from the beginning and end.

```python
text = "   Python   "

print(text.strip())
```

Output:

```text
Python
```

### `lstrip()`

Removes spaces from the left.

```python
text = "   Python"
print(text.lstrip())
```

### `rstrip()`

Removes spaces from the right.

```python
text = "Python   "
print(text.rstrip())
```

---

# 13. `replace()`

Replaces one string with another.

```python
text = "I like Java"

print(text.replace("Java", "Python"))
```

Output:

```text
I like Python
```

---

# 14. `find()`

Finds the position of a substring.

```python
text = "Python Programming"

print(text.find("Program"))
```

Output:

```text
7
```

If the text isn't found, `find()` returns `-1`.

# 18. `split()`

Splits a string into a list.

```python
text = "Python Java SQL"

result = text.split()

print(result)
```

Output:

```text
['Python', 'Java', 'SQL']
```

You can specify a separator:

```python
data = "Python,Java,SQL"

print(data.split(","))
```

Output:

```text
['Python', 'Java', 'SQL']
```

---

# 19. `join()`

`join()` combines elements into a string.

```python

result = " | ".join(languages)
```

Output:

```text

```

# 25. String Comparison

Strings can be compared using comparison operators.

```python
a = "apple"
b = "banana"

print(a == b)
print(a != b)
```

Output:

```text
False
True
```

Python compares strings lexicographically, broadly based on Unicode character values.

# 26. f-Strings

f-strings are one of the easiest ways to insert variables into strings.

```python
name = "Balkrushna"
age = 24

print(f"My name is {name} and I am {age} years old.")
```

Output:

```text
My name is Balkrushna and I am 24 years old.
```

# 27. Escape Characters

Escape characters allow special characters inside strings.

### New line: `\n`

```python
print("Hello\nPython")
```

Output:

```text
Hello
Python
```

### Tab: `\t`

```python
print("Name\tAge")
```

Output:

```text
Name    Age
```

### Double quote: `\"`

```python
print("He said, \"Hello\"")
```

### Backslash: `\\`

```python
print("C:\\Users\\Python")
```

---

# 28. Raw Strings

A raw string treats backslashes mostly as literal characters.

```python
path = r"C:\Users\Balkrushna\Python"

print(path)
```

This is especially useful for Windows file paths and regular expressions.

---

# 29. Strings Are Immutable

Strings cannot be changed directly after creation.

This will cause an error:

```python
text = "Python"

text[0] = "J"
```

Instead, create a new string:

```python
text = "Python"

text = "J" + text[1:]

print(text)
```

Output:

```text
Jython
```

# 32. Check Palindrome

A palindrome reads the same forward and backward.

```python
text = "madam"

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

Output:

```text
Palindrome
```

# 33. String Formatting

Another way to format strings is `.format()`.
