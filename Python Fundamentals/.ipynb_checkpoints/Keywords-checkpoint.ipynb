# Python Keywords

**Keywords** are reserved words in Python that have a special meaning to the Python interpreter. You **cannot use keywords as variable, function, or class names**.

For example:

```python
if = 10
```

This produces a **SyntaxError** because `if` is a Python keyword.

---

# 2. Conditional Keywords

Python uses `if`, `elif`, and `else` for decision-making.

```python
age = 20

if age >= 18:
    print("Adult")
elif age >= 13:
    print("Teenager")
else:
    print("Child")
```

Here:

* `if` → checks the first condition
* `elif` → checks another condition
* `else` → executes when previous conditions are false

---

# 3. Loop Keywords

### `for`

Used to iterate over a sequence.

```python
for i in range(5):
    print(i)
```

### `while`

Repeats code while a condition is true.

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

### `break`

Stops a loop.

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

Output:

```text
0
1
2
3
4
```

### `continue`

Skips the current iteration.

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

Output:

```text
0
1
3
4
```

### `pass`

Does nothing. It is commonly used as a placeholder.

```python
for i in range(5):
    pass
```

---

# 4. Function Keywords

### `def`

Used to define a function.

```python
def greet():
    print("Hello")

greet()
```

### `return`

Sends a value back from a function.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

Output:

```text
30
```

### `lambda`

Creates a small anonymous function.

```python
square = lambda x: x ** 2

print(square(5))
```

Output:

```text
25
```

# 7. Membership and Identity Keywords

### `in`

Checks membership:

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
```

### `is`

Checks object identity:

```python
a = None

print(a is None)
```

Output:

```text
True
```

---

# 8. Exception Handling Keywords

Python uses `try`, `except`, `else`, `finally`, and `raise` for handling errors.

```python
try:
    number = int(input("Enter a number: "))
    print(10 / number)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

### `raise`

You can manually raise an exception:

```python
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
```

---

# 9. Import Keywords

### `import`

Used to import a module.

```python
import math

print(math.sqrt(25))
```

### `from`

Used to import a specific item.

```python
from math import sqrt

print(sqrt(25))
```

### `as`

Used to create an alias.

```python
import pandas as pd
```

Now instead of:

```python
pandas.DataFrame()
```

you can write:

```python
pd.DataFrame()
```

# 10. `None`

`None` represents the absence of a value.

```python
result = None

print(result)
```

Output:

```text
None
```

It is commonly used when a variable currently has no value.

---

# 11. `class`

The `class` keyword is used to create a class.

```python
class Student:
    pass
```

Classes are used in **Object-Oriented Programming (OOP)**.

---

# 12. `with`

`with` is commonly used for resource management, especially files.

```python
with open("data.txt", "r") as file:
    data = file.read()
```

Python automatically handles closing the file after the block.

---

# 13. Check Keywords Using Python

You don't need to memorize the complete list. Python can tell you which words are keywords.

```python
import keyword

print(keyword.kwlist)
```

You can also check whether a specific word is a keyword:

```python
import keyword

print(keyword.iskeyword("if"))
print(keyword.iskeyword("name"))
```

Output:

```text
True
False
```

# 14. Keyword vs Variable

### ❌ Incorrect
