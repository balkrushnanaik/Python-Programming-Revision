# Conditional Statements in Python

**Conditional statements** in Python are used to make decisions in a program. They execute different blocks of code depending on whether a condition is `True` or `False`.

Python mainly provides:

* `if`
* `if-else`
* `if-elif-else`
* Nested `if`
* Conditional expression (ternary operator)

---

## 1. `if` Statement

The `if` statement executes a block of code only when the condition is `True`.

### Syntax

```python
if condition:
    # code to execute
```

### Example

```python
age = 20

if age >= 18:
    print("You are eligible to vote")
```

Output:

```text
You are eligible to vote
```

### Important

Python uses **indentation** to identify the code belonging to the `if` block.

```python
if age >= 18:
    print("Eligible")   # inside if
```

---

## 2. `if-else` Statement

The `else` block executes when the `if` condition is `False`.

### Syntax

```python
if condition:
    # code if condition is True
else:
    # code if condition is False
```

### Example

```python
age = 16

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

Output:

```text
Not eligible to vote
```

---

## 3. `if-elif-else` Statement

When you need to check **multiple conditions**, use `elif`.

`elif` means **else if**.

### Syntax

```python
if condition1:
    # code
elif condition2:
    # code
elif condition3:
    # code
else:
    # code
```

### Example

```python
marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Grade D")
```

Output:

```text
Grade B
```

Python checks the conditions **from top to bottom** and executes the first condition that is `True`.

---

## 4. Nested `if` Statement

An `if` statement inside another `if` statement is called a **nested if**.

### Example

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("ID required")
else:
    print("Entry not allowed")
```

Output:

```text
Entry allowed
```

---

## 5. Using Comparison Operators

Conditional statements commonly use comparison operators.

| Operator | Meaning               | Example   |
| -------- | --------------------- | --------- |
| `==`     | Equal to              | `x == 10` |
| `!=`     | Not equal to          | `x != 10` |
| `>`      | Greater than          | `x > 10`  |
| `<`      | Less than             | `x < 10`  |
| `>=`     | Greater than or equal | `x >= 10` |
| `<=`     | Less than or equal    | `x <= 10` |

Example:

```python
age = 25

if age > 18:
    print("Adult")
```

---

## 6. Using Logical Operators

You can combine multiple conditions using:

### `and`

Both conditions must be `True`.

```python
age = 25
salary = 50000

if age >= 18 and salary >= 30000:
    print("Eligible")
```

### `or`

At least one condition must be `True`.

```python
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("Weekend")
```

### `not`

Reverses the condition.

```python
is_raining = False

if not is_raining:
    print("Go outside")
```

---

## 7. Conditional Expression (Ternary Operator)

Python allows a simple `if-else` to be written in one line.

### Syntax

```python
value_if_true if condition else value_if_false
```

### Example

```python
age = 20

status = "Adult" if age >= 18 else "Minor"

print(status)
```

Output:

```text
Adult
```

---

## 8. Checking Multiple Conditions

Example:

```python
temperature = 35

if temperature >= 40:
    print("Very Hot")
elif temperature >= 30:
    print("Hot")
elif temperature >= 20:
    print("Normal")
else:
    print("Cold")
```

Output:

```text
Hot
```

---

## 9. Truthy and Falsy Values

Python also allows you to use values directly as conditions.

### Falsy values

These are generally treated as `False`:

```python
False
None
0
0.0
""
[]
()
{}
```

Example:

```python
name = ""

if name:
    print("Name is available")
else:
    print("Name is empty")
```

Output:

```text
Name is empty
```

---

## 10. Real-World Example

```python
username = "admin"
password = "1234"

if username == "admin" and password == "1234":
    print("Login successful")
else:
    print("Invalid username or password")
```

Output:

```text
Login successful
```

---

## Quick Summary

```text
if              → Check one condition
if-else         → Choose between two possibilities
if-elif-else    → Choose between multiple possibilities
nested if       → Put one condition inside another
ternary         → Write simple if-else in one line
```

### Example combining everything

```python
marks = 82

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print("Grade:", grade)
```

Output:

```text
Grade: A
```

**Key concept:** Conditional statements are the foundation of **decision-making in Python**. They are used extensively with loops, functions, user input, and data analysis.
