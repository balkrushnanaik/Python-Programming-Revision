# Loops in Python

**Loops** in Python are used to execute a block of code repeatedly. They are useful when you need to perform the same operation multiple times.

Python mainly has two types of loops:

1. `for` loop
2. `while` loop

Python also provides loop-control statements such as:

* `break`
* `continue`
* `pass`

---

## 1. `for` Loop

A `for` loop is used to iterate over a sequence such as a **list, tuple, string, set, dictionary, or range**.

### Syntax

```python
for variable in sequence:
    # code to execute
```

### Example

```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

Output:

```text
Apple
Banana
Mango
```

The loop takes each item from the list one by one.

---

# 2. `for` Loop with `range()`

The `range()` function is commonly used when you want to repeat something a specific number of times.

### Example

```python
for i in range(5):
    print(i)
```

Output:

```text
0
1
2
3
4
```

Remember: `range(5)` starts at `0` and stops **before** `5`.

### Starting and Ending Values

```python
for i in range(1, 6):
    print(i)
```

Output:

```text
1
2
3
4
5
```

### Using a Step

```python
for i in range(2, 11, 2):
    print(i)
```

Output:

```text
2
4
6
8
10
```

Here:

```text
range(start, stop, step)
```

---

# 3. Loop Through a String

A string is a sequence of characters, so you can iterate through it.

```python
name = "Python"

for char in name:
    print(char)
```

Output:

```text
P
y
t
h
o
n
```

---

# 4. Loop Through a List

```python
numbers = [10, 20, 30, 40, 50]

for number in numbers:
    print(number)
```

Output:

```text
10
20
30
40
50
```

You can also perform calculations:

```python
numbers = [10, 20, 30]

for number in numbers:
    print(number * 2)
```

Output:

```text
20
40
60
```

---

# 5. Loop Through a Dictionary

Dictionaries contain **key-value pairs**.

```python
student = {
    "name": "Rahul",
    "age": 21,
    "marks": 85
}
```

### Loop Through Keys

```python
for key in student:
    print(key)
```

Output:

```text
name
age
marks
```

### Loop Through Values

```python
for value in student.values():
    print(value)
```

Output:

```text
Rahul
21
85
```

### Loop Through Keys and Values

```python
for key, value in student.items():
    print(key, ":", value)
```

Output:

```text
name : Rahul
age : 21
marks : 85
```

---

# 6. `while` Loop

A `while` loop executes repeatedly **as long as a condition is `True`**.

### Syntax

```python
while condition:
    # code
```

### Example

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

Output:

```text
1
2
3
4
5
```

Here, `count += 1` is important because it eventually makes the condition `False`.

---

# 7. Infinite Loop

An infinite loop occurs when the condition never becomes `False`.

```python
count = 1

while count <= 5:
    print(count)
```

This keeps running because `count` never changes.

A common way to stop such a loop is using `break`.

---

# 8. `break` Statement

`break` immediately terminates the loop.

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

Output:

```text
1
2
3
4
```

When `i` becomes `5`, the loop stops.

---

# 9. `continue` Statement

`continue` skips the current iteration and moves to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

Output:

```text
1
2
4
5
```

The number `3` is skipped.

---

# 10. `pass` Statement

`pass` does nothing. It is used as a placeholder when you need a statement syntactically but do not want to execute anything yet.

```python
for i in range(5):
    pass
```

It is commonly useful while developing code that will be completed later.

---

# 11. Nested Loops

A loop inside another loop is called a **nested loop**.

### Example

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

Output:

```text
1 1
1 2
1 3
2 1
2 2
2 3
3 1
3 2
3 3
```

Nested loops are often used for working with tables, matrices, and patterns.

---

# 12. `else` with Loops

Python allows an `else` block with both `for` and `while` loops.

The `else` block runs when the loop completes **normally**, without being stopped by `break`.

```python
for i in range(5):
    print(i)
else:
    print("Loop completed")
```

Output:

```text
0
1
2
3
4
Loop completed
```

With `break`:

```python
for i in range(5):
    if i == 2:
        break
    print(i)
else:
    print("Loop completed")
```

Output:

```text
0
1
```

The `else` block does not execute because the loop was terminated by `break`.

---

# 13. Practical Example: Even Numbers

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```

Output:

```text
2
4
6
8
10
```

---

# 14. Practical Example: Sum of Numbers

```python
numbers = [10, 20, 30, 40, 50]

total = 0

for number in numbers:
    total += number

print("Total:", total)
```

Output:

```text
Total: 150
```

---

# 15. Practical Example: Multiplication Table

```python
number = 5

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

Output:

```text
5 x 1 = 5
5 x 2 = 10
5 x 3 = 15
5 x 4 = 20
5 x 5 = 25
5 x 6 = 30
5 x 7 = 35
5 x 8 = 40
5 x 9 = 45
5 x 10 = 50
```

---

# `for` vs `while`

| `for` Loop                          | `while` Loop                                   |
| ----------------------------------- | ---------------------------------------------- |
| Used to iterate over a sequence     | Used when repetition depends on a condition    |
| Often used with `range()`           | Continues while condition is `True`            |
| Number of iterations is often known | Number of iterations may be unknown            |
| Example: list traversal             | Example: running until user enters valid input |

### Example

```python
# for loop
for i in range(5):
    print(i)
```

```python
# while loop
i = 0

while i < 5:
    print(i)
    i += 1
```

Both produce:

```text
0
1
2
3
4
```

---

## Quick Summary

```text
for       → Iterate over a sequence
while     → Repeat while a condition is true
break     → Stop the loop
continue  → Skip the current iteration
pass      → Do nothing; placeholder
nested    → Loop inside another loop
range()   → Generate a sequence of numbers
```

### Simple example combining loops and conditions

```python
for number in range(1, 11):
    if number % 2 == 0:
        print(number, "is even")
    else:
        print(number, "is odd")
```

This demonstrates an important Python pattern:

**Loop → Check condition → Perform an action.**
