### 🐍 What is Python?

**Python is a high-level, interpreted, general-purpose programming language** known for its simple syntax, readability, and wide range of applications.

Python was created by **Guido van Rossum** and first released in **1991**.

### Simple Definition

> **Python is a programming language used to write instructions for computers, with a focus on simplicity and readability.**

### Key Features of Python

* **Easy to learn** – Simple and readable syntax.
* **High-level** – You don't need to manage low-level computer operations.
* **Interpreted** – Python code is executed by the Python interpreter.
* **Dynamically typed** – You don't have to explicitly declare variable types.
* **Object-oriented** – Supports classes and objects.
* **Cross-platform** – Works on Windows, macOS, and Linux.
* **Large ecosystem** – Thousands of libraries and frameworks are available.
* **Open-source** – Freely available to use and modify.
