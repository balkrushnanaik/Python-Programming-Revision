# Python Variables

A **variable** in Python is a name used to store a value. Think of a variable as a **label attached to some data**.

### 1. Creating Variables

Python does not require you to declare the variable type separately.

```python
name = "Balkrushna"
age = 25
salary = 50000.50
```

Here:

* `name` stores a string
* `age` stores an integer
* `salary` stores a floating-point number

You can display them using `print()`:

```python
print(name)
print(age)
print(salary)
```

---

### 2. Variable Assignment

The `=` operator is used to assign a value to a variable.

```python
x = 10
y = 20

result = x + y

print(result)
```

**Output:**

```text
30
```

---

### 3. Variables Can Change

A variable's value can be changed during program execution.

```python
age = 20
print(age)

age = 25
print(age)
```

**Output:**

```text
20
25
```

---

### 4. Different Data Types

Python variables can store different types of data.

```python
name = "Rahul"       # str
age = 25             # int
height = 5.9         # float
is_student = True    # bool
```

You can check the type using `type()`:

```python
print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

**Output:**

```text
<class 'str'>
<class 'int'>
<class 'float'>
<class 'bool'>
```

---

### 5. Multiple Variable Assignment

You can assign values to multiple variables in one line.

```python
x, y, z = 10, 20, 30

print(x)
print(y)
print(z)
```

You can also assign the **same value** to multiple variables:

```python
a = b = c = 100

print(a, b, c)
```

**Output:**

```text
100 100 100
```

---

### 6. Variable Naming Rules

Python variable names must follow certain rules.

#### ✅ Valid names

```python
name = "Alice"
age = 25
student_name = "Rahul"
_marks = 90
marks2 = 95
```

#### ❌ Invalid names

```python
2name = "Alice"       # Cannot start with a number
student-name = "Bob"  # Hyphen is not allowed
class = "Python"      # Keyword cannot be used
```

### Important Rules

1. A variable **cannot start with a number**.
2. It can start with a letter or `_`.
3. It can contain letters, numbers, and `_`.
4. Python is **case-sensitive**.
5. Don't use Python keywords as variable names.

For example:

```python
age = 20
Age = 30

print(age)
print(Age)
```

These are **two different variables**.

---

### 7. Python is Dynamically Typed

Python automatically determines the data type of a variable.

```python
x = 10
print(type(x))

x = "Hello"
print(type(x))
```

The same variable can refer to values of different types at different times.

---

### 8. Taking User Input into Variables

Variables are commonly used with `input()`.

```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Name:", name)
print("Age:", age)
```

**Sample:**

```text
Enter your name: Balkrushna
Enter your age: 25

Name: Balkrushna
Age: 25
```

---

### 9. Variable Naming Convention

Python commonly uses **snake_case** for variable names.

```python
first_name = "Rahul"
last_name = "Sharma"
total_marks = 450
employee_salary = 60000
```

Avoid names like:

```python
fn = "Rahul"
x1 = 450
abc = 60000
```

when a descriptive name is possible.

---

### Quick Summary

| Concept             | Example            |
| ------------------- | ------------------ |
| Create variable     | `x = 10`           |
| String              | `name = "John"`    |
| Integer             | `age = 25`         |
| Float               | `price = 99.99`    |
| Boolean             | `is_active = True` |
| Check type          | `type(x)`          |
| Multiple assignment | `a, b = 10, 20`    |
| Same value          | `a = b = 10`       |
| User input          | `name = input()`   |

**Key idea:** In Python, you simply write `variable_name = value`; Python determines the type automatically.
