# Python Operators

**Operators** are special symbols or keywords used to perform operations on values and variables.

For example:

```python
a = 10
b = 5

print(a + b)
```

**Output:**

```text
15
```

Python has several types of operators.

---

## 1. Arithmetic Operators

Arithmetic operators are used for mathematical calculations.

| Operator | Name           |   Example |     Result |
| -------- | -------------- | --------: | ---------: |
| `+`      | Addition       |  `10 + 3` |       `13` |
| `-`      | Subtraction    |  `10 - 3` |        `7` |
| `*`      | Multiplication |  `10 * 3` |       `30` |
| `/`      | Division       |  `10 / 3` | `3.333...` |
| `//`     | Floor Division | `10 // 3` |        `3` |
| `%`      | Modulus        |  `10 % 3` |        `1` |
| `**`     | Exponentiation | `10 ** 3` |     `1000` |

### Example

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
```

---

## 2. Comparison Operators

Comparison operators compare two values and return either `True` or `False`.

| Operator | Meaning                  | Example    |
| -------- | ------------------------ | ---------- |
| `==`     | Equal to                 | `10 == 10` |
| `!=`     | Not equal to             | `10 != 5`  |
| `>`      | Greater than             | `10 > 5`   |
| `<`      | Less than                | `10 < 5`   |
| `>=`     | Greater than or equal to | `10 >= 10` |
| `<=`     | Less than or equal to    | `5 <= 10`  |

### Example

```python
age = 25

print(age == 25)
print(age > 18)
print(age < 18)
print(age != 30)
```

**Output:**

```text
True
True
False
True
```

---

## 3. Assignment Operators

Assignment operators are used to assign and update values.

| Operator | Example   | Equivalent   |
| -------- | --------- | ------------ |
| `=`      | `x = 10`  | `x = 10`     |
| `+=`     | `x += 5`  | `x = x + 5`  |
| `-=`     | `x -= 5`  | `x = x - 5`  |
| `*=`     | `x *= 5`  | `x = x * 5`  |
| `/=`     | `x /= 5`  | `x = x / 5`  |
| `//=`    | `x //= 5` | `x = x // 5` |
| `%=`     | `x %= 5`  | `x = x % 5`  |
| `**=`    | `x **= 5` | `x = x ** 5` |

### Example

```python
x = 10

x += 5
print(x)

x *= 2
print(x)
```

**Output:**

```text
15
30
```

---

## 4. Logical Operators

Logical operators are used to combine conditions.

Python has three logical operators:

### `and`

Returns `True` if **both conditions are True**.

```python
age = 25
salary = 50000

print(age > 18 and salary > 30000)
```

**Output:**

```text
True
```

### `or`

Returns `True` if **at least one condition is True**.

```python
age = 16
has_permission = True

print(age >= 18 or has_permission)
```

**Output:**

```text
True
```

### `not`

Reverses the Boolean result.

```python
is_student = True

print(not is_student)
```

**Output:**

```text
False
```

---

## 5. Identity Operators

Identity operators check whether two variables refer to the **same object**.

| Operator | Meaning           |
| -------- | ----------------- |
| `is`     | Same object       |
| `is not` | Different objects |

Example:

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b)
print(a is c)
print(a is not c)
```

**Output:**

```text
True
False
True
```

### `is` vs `==`

This is important:

```python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)
print(a is b)
```

**Output:**

```text
True
False
```

* `==` checks whether the **values are equal**.
* `is` checks whether they are the **same object**.

---

## 6. Membership Operators

Membership operators check whether a value exists inside a sequence such as a string, list, tuple, etc.

| Operator | Meaning           |
| -------- | ----------------- |
| `in`     | Exists in         |
| `not in` | Does not exist in |

### Example

```python
fruits = ["apple", "banana", "mango"]

print("apple" in fruits)
print("orange" in fruits)
print("orange" not in fruits)
```

**Output:**

```text
True
False
True
```

With strings:

```python
name = "Python"

print("P" in name)
print("Java" in name)
```

---

## 7. Bitwise Operators

Bitwise operators work with numbers at the **binary/bit level**.

| Operator | Name        |    |
| -------- | ----------- | -- |
| `&`      | AND         |    |
| `        | `           | OR |
| `^`      | XOR         |    |
| `~`      | NOT         |    |
| `<<`     | Left Shift  |    |
| `>>`     | Right Shift |    |

Example:

```python
a = 5
b = 3

print(a & b)
print(a | b)
print(a ^ b)
```

Binary representation:

```text
5 = 101
3 = 011
```

Therefore:

```text
5 & 3 = 001 = 1
5 | 3 = 111 = 7
5 ^ 3 = 110 = 6
```

---

# 8. Operator Precedence

When an expression contains multiple operators, Python follows **operator precedence**.

For example:

```python
result = 10 + 5 * 2

print(result)
```

**Output:**

```text
20
```

Not `30`, because multiplication happens before addition.

Using parentheses:

```python
result = (10 + 5) * 2

print(result)
```

**Output:**

```text
30
```

### Common precedence order

From higher to lower:

1. `()` — Parentheses
2. `**` — Exponentiation
3. `*`, `/`, `//`, `%`
4. `+`, `-`
5. Comparison operators
6. `not`
7. `and`
8. `or`

---

# 9. Combining Operators

You can use multiple operators in a single expression.

```python
marks = 85
attendance = 90

result = marks >= 40 and attendance >= 75

print(result)
```

**Output:**

```text
True
```

Another example:

```python
age = 25
salary = 50000

eligible = age >= 21 and salary >= 30000

print(eligible)
```

**Output:**

```text
True
```

## Quick Summary

| Category   | Operators                  |
| ---------- | -------------------------- |
| Arithmetic | `+ - * / // % **`          |
| Comparison | `== != > < >= <=`          |
| Assignment | `= += -= *= /= //= %= **=` |
| Logical    | `and or not`               |
| Identity   | `is`, `is not`             |
| Membership | `in`, `not in`             |
| Bitwise    | `& \| ^ ~ << >>`           |

### ⭐ Most Important for Beginners

Focus first on:

```text
Arithmetic → Comparison → Assignment → Logical → Membership
```

These operators are used constantly when writing Python programs, especially with **if-else statements, loops, functions, and data analysis**.
