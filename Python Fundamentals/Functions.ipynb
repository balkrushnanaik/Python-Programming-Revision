# Python Functions

A **function** in Python is a reusable block of code that performs a specific task. Functions help make programs **organized, reusable, readable, and easier to maintain**.

## 1. Basic Syntax

```python
def function_name():
    # code
```

Example:

```python
def greet():
    print("Hello, World!")

greet()
```

Output:

```text
Hello, World!
```

Here:

* `def` → keyword used to define a function
* `greet` → function name
* `()` → parameters go here
* `greet()` → function call

---

## 2. Function Without Parameters

```python
def welcome():
    print("Welcome to Python!")

welcome()
```

---

## 3. Function With Parameters

Parameters allow us to pass data into a function.

```python
def greet(name):
    print("Hello", name)

greet("Balkrushna")
```

Output:

```text
Hello Balkrushna
```

Another example:

```python
def add(a, b):
    print(a + b)

add(10, 20)
```

Output:

```text
30
```

---

## 4. Function With Return Value

`return` sends a result back to the calling code.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

Output:

```text
30
```

Difference:

```python
print(a + b)
```

displays the result, whereas:

```python
return a + b
```

returns the result so it can be stored or used elsewhere.

---

## 5. Multiple Parameters

```python
def student_info(name, age, city):
    print("Name:", name)
    print("Age:", age)
    print("City:", city)

student_info("Balkrushna", 24, "Pune")
```

---

## 6. Positional Arguments

Arguments are matched according to their position.

```python
def details(name, age):
    print(name, age)

details("Rahul", 22)
```

Here:

* `"Rahul"` → `name`
* `22` → `age`

---

## 7. Keyword Arguments

You can specify the parameter name while calling the function.

```python
def details(name, age):
    print(name, age)

details(age=22, name="Rahul")
```

Output:

```text
Rahul 22
```

---

## 8. Default Arguments

A parameter can have a default value.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Balkrushna")
```

Output:

```text
Hello Guest
Hello Balkrushna
```

---

## 9. `*args`

`*args` allows a function to accept multiple positional arguments.

```python
def add_numbers(*numbers):
    total = 0

    for number in numbers:
        total += number

    return total

print(add_numbers(10, 20))
print(add_numbers(10, 20, 30, 40))
```

Output:

```text
30
100
```

---

## 10. `**kwargs`

`**kwargs` allows multiple keyword arguments.

```python
def student(**details):
    print(details)

student(name="Rahul", age=22, city="Pune")
```

Output:

```text
{'name': 'Rahul', 'age': 22, 'city': 'Pune'}
```

---

## 11. Local Variable

A variable created inside a function is generally accessible only inside that function.

```python
def test():
    x = 10
    print(x)

test()
```

---

## 12. Global Variable

A variable defined outside a function has global scope.

```python
x = 100

def test():
    print(x)

test()
```

Output:

```text
100
```

---

## 13. Using `global`

You can modify a global variable using `global`.

```python
x = 10

def change():
    global x
    x = 50

change()

print(x)
```

Output:

```text
50
```

---

## 14. Function With Conditions

```python
def check_number(num):
    if num > 0:
        return "Positive"
    elif num < 0:
        return "Negative"
    else:
        return "Zero"

print(check_number(10))
```

Output:

```text
Positive
```

---

## 15. Function With a Loop

```python
def print_numbers(n):
    for i in range(1, n + 1):
        print(i)

print_numbers(5)
```

Output:

```text
1
2
3
4
5
```

---

## 16. Recursive Function

A function calling itself is called a **recursive function**.

Example: factorial

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

Output:

```text
120
```

---

## 17. Lambda Function

A **lambda** is a small anonymous function.

```python
square = lambda x: x * x

print(square(5))
```

Output:

```text
25
```

Equivalent normal function:

```python
def square(x):
    return x * x
```

---

## 18. Type Hints

Python allows you to specify expected parameter and return types.

```python
def add(a: int, b: int) -> int:
    return a + b

print(add(10, 20))
```

Type hints improve readability but generally do not enforce the types at runtime.

---

## 19. Docstring

A docstring explains what a function does.

```python
def add(a, b):
    """Returns the sum of two numbers."""
    return a + b

print(add(10, 20))
```

You can view it using:

```python
print(add.__doc__)
```

# Important Function Concepts

| Concept          | Meaning                          |
| ---------------- | -------------------------------- |
| `def`            | Defines a function               |
| Function call    | Executes a function              |
| Parameter        | Variable in function definition  |
| Argument         | Actual value passed to function  |
| `return`         | Sends a value back               |
| Default argument | Parameter with a default value   |
| `*args`          | Multiple positional arguments    |
| `**kwargs`       | Multiple keyword arguments       |
| Recursion        | Function calls itself            |
| Lambda           | Anonymous function               |
| Scope            | Where a variable can be accessed |
| Docstring        | Function documentation           |

## Simple Real-World Example

Suppose you're building a **Data Analytics application**:

```python
def calculate_average(marks):
    total = sum(marks)
    average = total / len(marks)
    return average

marks = [75, 80, 65, 90, 85]

result = calculate_average(marks)

print("Average Marks:", result)
```

Output:

```text
Average Marks: 79.0
```

This is one of the most important reasons functions are useful in data analytics: you can write the calculation **once and reuse it for thousands of records**.

### Functions learning sequence

**Basic function → Parameters → Arguments → Return → Default arguments → Keyword arguments → `*args` → `**kwargs` → Scope → Recursion → Lambda → Built-in functions → Higher-order functions**
