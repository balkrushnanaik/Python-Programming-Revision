In Python, **input** is used to get data from the user, and **output** is used to display data on the screen.

### 1. Output in Python

The `print()` function is used to display output.

**Example:**

```python
print("Hello, World!")
```

**Output:**

```
Hello, World!
```

You can also print variables:

```python
name = "Alice"
print("Name:", name)
```

**Output:**

```
Name: Alice
```

---

### 2. Input in Python

The `input()` function is used to take input from the user.

**Example:**

```python
name = input("Enter your name: ")
print("Hello,", name)
```

**Sample Output:**

```
Enter your name: John
Hello, John
```

---

### 3. Input of Numbers

By default, `input()` returns a string. To use numbers, convert the input using `int()` or `float()`.

**Integer Example:**

```python
age = int(input("Enter your age: "))
print("Your age is", age)
```

**Float Example:**

```python
price = float(input("Enter the price: "))
print("Price:", price)
```

---

### 4. Multiple Inputs

You can take multiple inputs in one line using `split()`.

```python
a, b = map(int, input("Enter two numbers: ").split())
print("Sum =", a + b)
```

**Sample Input:**

```
10 20
```

**Output:**

```
Sum = 30
```

---

### Summary

* `print()` → Displays output.
* `input()` → Takes input from the user.
* `int()` → Converts input to an integer.
* `float()` → Converts input to a decimal number.
* `split()` → Splits multiple inputs entered on one line.

These functions form the foundation of user interaction in Python programs.
