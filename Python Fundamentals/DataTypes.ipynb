# Python Data Types

In Python, **data types** specify what kind of value a variable stores. Python automatically determines the data type when you assign a value.

### 1. Numeric Data Types

#### `int` — Integer

Stores whole numbers.

```python
age = 25
marks = 90
```

```python
print(type(age))
# <class 'int'>
```

#### `float` — Decimal Number

Stores numbers with decimal points.

```python
price = 99.50
height = 5.11
```

#### `complex` — Complex Number

Stores complex numbers.

```python
z = 3 + 4j
print(z)
# (3+4j)
```

---

### 2. Boolean Data Type

#### `bool`

Stores either `True` or `False`.

```python
is_student = True
is_logged_in = False
```

Example:

```python
age = 20
print(age >= 18)
# True
```

---

### 3. String Data Type

#### `str`

Stores text or a sequence of characters.

```python
name = "Balkrushna"
message = 'Hello Python'
```

You can access individual characters:

```python
name = "Python"

print(name[0])
# P

print(name[-1])
# n
```

---

### 4. List

#### `list`

Stores multiple values in an **ordered and changeable** collection.

```python
fruits = ["Apple", "Banana", "Mango"]
```

You can modify a list:

```python
fruits[1] = "Orange"

print(fruits)
# ['Apple', 'Orange', 'Mango']
```

---

### 5. Tuple

#### `tuple`

Stores multiple values in an **ordered but unchangeable** collection.

```python
colors = ("Red", "Green", "Blue")
```

Tuples are immutable:

```python
# colors[0] = "Yellow"  # Error
```

---

### 6. Set

#### `set`

Stores multiple **unique** values. Sets are unordered.

```python
numbers = {1, 2, 3, 4, 4}

print(numbers)
# {1, 2, 3, 4}
```

Duplicate values are automatically removed.

---

### 7. Dictionary

#### `dict`

Stores data in **key-value pairs**.

```python
student = {
    "name": "Rahul",
    "age": 21,
    "marks": 85
}
```

Access a value using its key:

```python
print(student["name"])
# Rahul
```

---

### 8. None Type

#### `NoneType`

Represents the absence of a value.

```python
result = None

print(type(result))
# <class 'NoneType'>
```

---

## Python Data Types Summary

| Data Type  | Example           | Description           |
| ---------- | ----------------- | --------------------- |
| `int`      | `10`              | Whole number          |
| `float`    | `10.5`            | Decimal number        |
| `complex`  | `2 + 3j`          | Complex number        |
| `bool`     | `True`            | True/False            |
| `str`      | `"Hello"`         | Text                  |
| `list`     | `[1, 2, 3]`       | Ordered, changeable   |
| `tuple`    | `(1, 2, 3)`       | Ordered, unchangeable |
| `set`      | `{1, 2, 3}`       | Unique values         |
| `dict`     | `{"name": "Raj"}` | Key-value pairs       |
| `NoneType` | `None`            | No value              |

### Checking the Data Type

Use the `type()` function:

```python
x = 100
y = 10.5
name = "Python"

print(type(x))
print(type(y))
print(type(name))
```

Output:

```text
<class 'int'>
<class 'float'>
<class 'str'>
```

**Important:** Python is a **dynamically typed language**, so you don't need to declare the data type explicitly.

```python
x = 10        # int
x = "Hello"   # now str
```

This is one of the important concepts to understand before learning Python collections and functions.
